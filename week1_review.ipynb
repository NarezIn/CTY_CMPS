## Variable Naming

### A Python variable ...
- cannot start with numbers (e.g. 1_student).
- cannot contain any special characters, except underscore (e.g. num*)
    - meaning can only contain alphanumeric characters and underscores (A-Z, a-z, 0-9, and _ )
- cannot be Python keywords. (e.g. if, else, while, for, elif...)

### The Convention (recommended but not required):
- A Python variable should be written in snake_case.
    - meaning all lowercase letters with words separated by underscores
    e.g. you_will_have_exam_today, student_score, total_price

## Review

- range() function can accept 1 to 3 arguments, and it returns a sequence of number depending on the arguments.
    - if three arguments are passed, the first argument would be the start, the second argument would be the end, and the third argument would be the step_size. (range(1, 9, 3))

    - if you only pass one argument to range(), that argument would be the stop. So the start would be 0 and the step_size would be 1 in default. (e.g. range(5))
    
    - if you pass two arguments to range(), the first argument would be the start and the second argument would be the end. The step_size would be 1 in default as mentioned above. (e.g. range(4, 10))



> Question: if I run the code sequence = range(5), will sequence have 5 in it?

sequence would be a sequence consists of 0, 1, 2, 3, 4

- for loop: for i in any_sequence. any_sequence can be list, tuple (you will learn later), dictionary, or even your own iterable class!
    - In each iteration (run) of the for loop, i would be an item in any_sequence, and the code inside the for loop that are indented would be run. After the indented code inside the for loop finishes running, i would be incremented (moved) to the next item in any_sequence
    - The for loop ends when i has no items to be in any_sequence.

```python
end = int(input("Pick an integer: "))
for i in range(end):
    print("Hi! Right now i is", i)
```

```python
# output?
start = 3
end = 9
step_size = 3
for i in range(start, end, step_size):
    print(i)
```

- Let's go over this for loop together.

    - range(3, 9, 3) returns a ...

    - In the first iteration, i would be 3. Then the indented code inside this loop would be executed: print(3).

    - In the second iteration, i would be 6. Then the indented code inside this loop would be executed: print(6).
    
    - Since 6 is the last item in the sequence, this for loops ends.

```python
for i in (1, 2, 3):
    print(i)
```

### You can do a lot of things with for loop.

```python
# e.g. Counting the number of a certain number that a sequence have.
# In this example, the sequence is a list, which we are all familiar with.
list_1 = [2, 2, 5, 10000, 2, 78, 3, 2]

counter = 0 # this variable helps me remember how many 2s would have in list_1
for i in range(len(list_1)):
    print("Now we are at index", i)

    if list_1[i] == 2:
        counter += 1 # this syntax is the same counter = counter + 1
        print("This number is 2.")
    else:
        print("This number is not 2.")
print("\nNumber of 2s in list_1:", counter)
```

```python
# I want to calculate the sum of all even numbers in a sequence.

# This chunk of code is for users to create a list
total_num = int(input("How many numbers do you want to put into a list? "))
list_2 = []
for i in range(total_num):
    choice = int(input("Pick a number: "))
    list_2.append(choice)
print("The list you created: ", list_2)

# This chunk of code is to calculate the sum of all even numbers in
# the list that the user creates above
even_sum = 0
for num in list_2:
    if num % 2 == 0:
        even_sum += num
print("The sum of all even numbers in the list you created is:", even_sum)
```

## String indexing and slicing

### Strings are somehow similar to lists. You can do indexing and slicing on them.

#### List indexing and slicing

```python
first_row = ["Zain", "Sidd", "Sebastian", "Logan", "Cj", "Alexis"]

white_shirt_student_there = first_row[3]

print(white_shirt_student_there)
```

```python
duo = first_row[4:]


print(duo)
```

```python
uno_or_duo = first_row[4:6]
print(duo == uno_or_duo)

# same logic as range(4, 5)
```
