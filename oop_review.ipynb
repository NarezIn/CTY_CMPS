# Object-Oriented-Programming (OOP)

- organize your code into reusable classes, instead of a sequential lines of steps (how we program before)

## OOP in Python: class
- treat `class` as the blueprint of objects.


```python
"""Let's import our dependencies at the top of the notebook/script."""
import random
in_the_wild = ["wooden stick", "tennis ball", "flying disc", "rubber chew toy"]
```

```python
class Dog:
    """
    This is the blueprint for all the actual dogs in the world.
    """
    def __init__(self, name, breed, age):
        self.name = name
        self.breed = breed
        self.age = age
        self.face = ""

    def bark(self):
        this_breed = self.breed.lower()
        if this_breed == "basenji":
            print("...")
        elif this_breed == "husky":
            print("woo-woo!~")
        elif this_breed == "chihuahua":
            print("yip yip!")
        else:
            print("woof!")

    def fetch(self):
        return random.choice(in_the_wild)

    def aging(self, year):
        self.age = self.age + year
        #self.age += year

    def introduce(self):
        print("Translation from dog language:\nHi, my name is", self.name + ". I'm a", str(self.age) + "-year-old", self.breed + ".")

    def bark_at(self, another_dog):
        print(self.name, "barks at", "a " + another_dog.breed + ". That does not look good.")
```

```python
# object1
dog_1 = Dog("Bobby", "Chihuahua", 2)

dog_1.bark()

print("Before aging:", dog_1.age)
dog_1.aging(2)
print("After aging:", dog_1.age)

item = dog_1.fetch()
print(dog_1.name, "fetches me a", item + ".\n")

dog_1.introduce()

print()
# object2
dog_2 = Dog("Lua", "German Shepherd", 5)
dog_2.bark()

print()
dog_1.bark_at(dog_2)
```
